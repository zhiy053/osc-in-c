# OSC

## 一个小Demo

```c
#include <stdio.h>
#include <unistd.h>
#include <sys/types.h>
#include <stdlib.h>

int main() {
    pid_t pid = 0;
    pid = fork();

    if (pid < 0) {
        printf("Could not create process\n");
        exit(1);
    } else if (pid == 0) {
        sleep(1);
        printf("Hello from the child process\n");
    } else if (pid > 0) {
        printf("Hello from the parent process\n");
    }

    printf("This code will be executed by both the child and parent\n");

    return 0;
}
```
